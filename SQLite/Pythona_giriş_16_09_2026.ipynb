```python
class Araba:
  def __init__(bu,marka,model,hiz):
    bu.marka = marka
    bu.model = model
    bu.hiz = hiz

  def hizlandir(bu,hiz):
     bu.hiz += hiz

  def bilgi_goster(bu):
    print(bu.marka,bu.model,bu.hiz)

araba = Araba("Honda","NSX",250)

araba.hizlandir(30)
araba.bilgi_goster()
```

```python
class Dikdortgen:
  def __init__(bu,uzunluk,genislik):
    bu.uzunluk = uzunluk
    bu.genislik = genislik

def alan_hesapla(bu):
  return bu.uzunluk*bu.genislik

def cevre_hesapla(bu):
  return 2*(bu.uzunluk+bu.genislik)

dikdortgen = Dikdortgen(5,4)
print(alan_hesapla(dikdortgen))
print(cevre_hesapla(dikdortgen))
```

```python
class Kitap:
  def __init__(bu,ad,yazar,sayfa_sayisi):
    bu.ad = ad
    bu.yazar = yazar
    bu.sayfa_sayisi = sayfa_sayisi

  def ozet(bu):
    return f"{bu.ad} - {bu.yazar} ({bu.sayfa_sayisi})"

kitap = Kitap("Suç ve Ceza","Dostoyevski",458)
print(kitap.ozet())
```

```python
class Urun:
  def __init__(self,isim,fiyat,stok_adedi):
    self.isim = isim
    self.fiyat = fiyat
    self.stok_adedi = stok_adedi

  def toplam_hesapla(self):
    return self.fiyat * self.stok_adedi

  def stok_durumu(self):
    if self.stok_adedi == 0:
      return "Tükendi"
    elif self.stok_adedi < 10:
      return "Stok azalmis"
    else:
      return "Yeterli stok"
```

```python
urun1 = Urun("Kulaklık",150,18)
print(f"{urun1.isim} toplam deger: {urun1.toplam_hesapla()}")
print(f"{urun1.isim} stok durumu: {urun1.stok_durumu()}")
```
